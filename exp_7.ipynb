```python
#7.3: Factorial of a number

n = int(input("Enter your Number: "))
fact = 1

while n > 0:
  fact *= n
  n -= 1

print ("Factorial of your Number is: ",fact)
```

```python
#7.4: Fibonacci Series using while Loop

n = int(input("Enter the number of terms: "))

a = 0
b = 1
i = 1

while i <= n:
  print(a, end=" ")
  c = a + b
  a = b
  b = c
  i += 1
```

```python
#7.5: Fibonacci Series using while Loop with Limits

limit = int(input("Enter the Limit: "))

a = 0
b = 1

while a <= limit:
  print(a, end=" ")
  a, b = b, a + b
```

```python
#7.6: Reverse the Number

num = int(input("Enter a Number: "))
rev = 0

while num > 0:
  digit = num % 10
  rev = rev * 10 + digit
  num //= 10

print("Reversed Number =", rev)
```

```python
#7.7 Check Palindrome Number using while loop

num = int(input("Enter a Number: "))
temp = num
rev = 0

while num > 0:
  rev = rev * 10 + num % 10
  num //= 10
if temp == rev:
  print ("Given Number is a Palindrome")
else:
  print ("Given Number is not a Palindrome")
```

```python
#7.8 Check Palindrome String using while loop

s = str(input("Enter a String: "))
i, j = 0, len(s) - 1
is_palindrome = True

while i < j:
  if s[i] != s[j]:
    is_palindrome = False
    break

  i += 1
  j -= 1

if is_palindrome:
  print("Given String is a Palindrome")
else:
  print("Given String is not a Palindrome")
```

```python
#7.9: Check Palindrome Number/String using if loop

st=input("Enter the String/Number: ")
rev = st[::-1]
if st == rev:
  print("Given String is a Palindrome")
else:
  print("Given String is not a Palindrome")
```

```python
#7.10: Count digits in a number using while loop

num = int(input("Enter a Number: "))
count = 0
while num > 0:
  count += 1
  num //= 10

print("Number of Digits in Given Number =", count)
```

```python
#7.12: Search an element in a List

nums = [10,20,30,40,50]
key = int(input("Enter Element to Search: "))
i = 0
while i < len(nums):
  if nums [i] == key:
    print("Element Found at Index",i)
    break
  i += 1
else:
  print("Element not Found")
```
